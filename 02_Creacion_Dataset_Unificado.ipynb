```python
import pandas as pd

# Cargar las tres bases de datos limpias
df_datd = pd.read_csv("./Bases de Datos Limpias/1 TheDATD/DATD_limpio.csv")
df_sdcnl = pd.read_csv("./Bases de Datos Limpias/2 SDCNL/SDCNL_limpio.csv")
df_depressionx = pd.read_csv("./Bases de Datos Limpias/3 DepressionX/DepressionX_limpio.csv")
```

```python
print("--- Etiquetas DATD ---")
print(df_datd['Label'].value_counts())

print("\n--- Etiquetas SDCNL ---")
print(df_sdcnl['Label'].value_counts())

print("\n--- Etiquetas DepressionX ---")
print(df_depressionx['Label'].value_counts())
```

```python
columnas_esperadas = ['Text', 'Label', 'Source']

print("DATD:", list(df_datd.columns))
print("SDCNL:", list(df_sdcnl.columns))
print("DepressionX:", list(df_depressionx.columns))

assert list(df_datd.columns) == columnas_esperadas
assert list(df_sdcnl.columns) == columnas_esperadas
assert list(df_depressionx.columns) == columnas_esperadas

print("\nTodas las bases de datos tienen las mismas columnas.")
```

```python
# Unir las tres bases de datos
Dataset_Unificado = pd.concat(
    [df_datd, df_sdcnl, df_depressionx],
    ignore_index=True
)

# Mezclar aleatoriamente las instancias
Dataset_Unificado = Dataset_Unificado.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)
```

```python
print("\n--- Información del Dataset Unificado ---")
print(Dataset_Unificado.info())

print("\n--- Conteo de clases del Dataset Unificado ---")
print(Dataset_Unificado['Label'].value_counts())

print("\n--- Conteo por fuente ---")
print(Dataset_Unificado['Source'].value_counts())

print("\n--- Conteo cruzado Source x Label ---")
print(pd.crosstab(Dataset_Unificado['Source'], Dataset_Unificado['Label']))

print("\n--- Primeros registros ---")
print(Dataset_Unificado.head())
```

```python
Dataset_Unificado.to_csv(
    "./Bases de Datos Limpias/4 Dataset_Unificado/Dataset_Unificado.csv",
    index=False,
    encoding='utf-8-sig'
)
```
